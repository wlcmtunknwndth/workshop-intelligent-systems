```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('house-votes-84.data', header=None)
df.columns = ['Class', 'handicapped-infants', 'water-project-cost-sharing', 'adoption-of-the-budget-resolution', 'physician-fee-freeze', 'el-salvador-aid', 'religious-groups-in-schools', 'anti-satellite-test-ban', 'aid-to-nicaraguan-contras', 'mx-missile', 'immigration', 'synfuels-corporation-cutback', 'education-spending', 'superfund-right-to-sue', 'crime', 'duty-free-exports', 'export-administration-act-south-africa']
df
```

## Балансировка классов

```python
df['Class'].value_counts()
```

```python
from imblearn.over_sampling import RandomOverSampler

x = df.drop('Class', axis = 1)
y = df['Class']

ros = RandomOverSampler(random_state=42)
x_new, y_new = ros.fit_resample(x, y)

y_new.value_counts()
```

```python
df_new = pd.DataFrame(x_new, columns=x.columns)  
df_new['Class'] = y_new
df_new
```

## Уникальные значения

```python
df_new = df_new.drop('export-administration-act-south-africa', axis = 1)
```

```python
df_new
```

## Кодировка

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import LabelEncoder

x2 = df_new.drop('Class', axis = 1)
y2 = df_new['Class']

encoder = OneHotEncoder(sparse_output=False)
encoded_array = encoder.fit_transform(x2)

l_encoder = LabelEncoder()
y_encoded = l_encoder.fit_transform(y2)

df_encode = pd.DataFrame(encoded_array, columns=encoder.get_feature_names_out(x2.columns))
df_encode['Class'] = y_encoded
df_encode
```

## Деление данных

```python
from sklearn.model_selection import train_test_split

x_n = df_encode.drop('Class', axis = 1)
y_n = df_encode['Class']

x_train, x_test, y_train, y_test = train_test_split(x_n, y_n, test_size=0.2)
x_train.shape, x_test.shape, y_train.shape, y_test.shape
```

## Логистическая регрессия

```python
from sklearn.linear_model import LogisticRegression

log_reg = LogisticRegression()
log_reg.fit(x_train, y_train)

log_reg.score(x_test, y_test)
importances = log_reg.coef_[0]
feature_names = log_reg.feature_names_in_
for feature, coef in zip(feature_names, importances):
    print(f"{feature}: {coef:.4f}")
```

```python
from sklearn.metrics import f1_score

y_pred = log_reg.predict(x_test)
f1_score(y_test, y_pred)
```

```python
from sklearn.metrics import accuracy_score

accuracy_score(y_test, y_pred)
```

```python
from sklearn.metrics import precision_score

precision_score(y_test, y_pred)
```

```python
from sklearn.metrics import recall_score

recall_score(y_test, y_pred)
```

## Метод ближайших соседей

```python
original_columns = x_train.columns
```

```python
from sklearn.neighbors import KNeighborsClassifier

x_train = x_train.to_numpy().astype(np.float64)
x_test = x_test.to_numpy().astype(np.float64)

min_err = 1.0
best_k = -1
for k in range(1, 11):
    knn_clf = KNeighborsClassifier(n_neighbors=k)
    knn_clf.fit(x_train, y_train)
    y_pred_k = knn_clf.predict(x_test)
    err = np.mean(y_pred_k!=y_test)
    if err < min_err:
        best_k = k
        min_err = err
        
print("Лучшее k =", best_k)
print("Минимальная доля ошибок =", min_err)
```

```python
from sklearn.inspection import permutation_importance


knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(x_train, y_train)
y_pred_knn = knn.predict(x_test)

# result = permutation_importance(knn, x_train, y_train, n_repeats=10, random_state=42)

# # Вывод результатов
# for i, importance in enumerate(result.importances_mean):
#     print(importance)

result = permutation_importance(knn, x_train, y_train, n_repeats=10, random_state=42)

# Выводим результаты
for i, col in enumerate(original_columns):
    print(f"{col}: {result.importances_mean[i]:.4f} ± {result.importances_std[i]:.4f}")
```

```python
f1_score(y_test, y_pred_knn)
```

```python
accuracy_score(y_test, y_pred_knn)
```

```python
precision_score(y_test, y_pred_knn)
```

```python
recall_score(y_test, y_pred_knn)
```

```python
x_train = pd.DataFrame(x_train, columns=original_columns)
x_test = pd.DataFrame(x_test, columns=original_columns)
```

## Наивный байесовский классификатор

```python
from sklearn.naive_bayes import GaussianNB

nbc = GaussianNB()
nbc.fit(x_train,y_train)
y_pred_nbc = nbc.predict(x_test)

means = pd.DataFrame(nbc.theta_, columns=original_columns, index=nbc.classes_)
variances = pd.DataFrame(nbc.var_, columns=original_columns, index=nbc.classes_)

# Оформленный вывод средних значений
print("Средние значения признаков для каждого класса:")
print(means)

print("\nДисперсии признаков для каждого класса:")
print(variances)
```

```python
f1_score(y_test, y_pred_nbc)
```

```python
accuracy_score(y_test, y_pred_nbc)
```

```python
precision_score(y_test, y_pred_nbc)
```

```python
recall_score(y_test, y_pred_nbc)
```

## Кросс-валидация

```python
x_n = x_n.to_numpy()
```

```python
from sklearn.model_selection import cross_validate

def cross_valid_func(model, x_n, y_n):
    scoring = ['accuracy', 'precision', 'recall', 'f1']
    scores = cross_validate(model, x_n, y_n, cv=5, scoring=scoring)

    print("Accuracy:", scores['test_accuracy'])
    print("Precision:", scores['test_precision'])
    print("Recall:", scores['test_recall'])
    print("F1-Score:", scores['test_f1'])

    print("\nСредние значения по всем фолдам:")
    print(f"Средняя Accuracy: {scores['test_accuracy'].mean():.4f}")
    print(f"Средняя Precision: {scores['test_precision'].mean():.4f}")
    print(f"Средняя Recall: {scores['test_recall'].mean():.4f}")
    print(f"Средняя F1-Score: {scores['test_f1'].mean():.4f}")
```

```python
cross_valid_func(log_reg, x_n, y_n)
```

```python
cross_valid_func(knn, x_n, y_n)
```

```python
cross_valid_func(nbc, x_n, y_n)
```

## Графики

```python
y_log_prob = log_reg.predict_proba(x_test)[:, 1]
y_nbc_prob = nbc.predict_proba(x_test)[:, 1]
```

```python
x_test = x_test.to_numpy().astype(np.float64)
y_knn_prob = knn.predict_proba(x_test)[:, 1]
```

```python
from sklearn.metrics import roc_curve, auc
import matplotlib.pyplot as plt

fp_knn, tp_knn, _ = roc_curve(y_test, y_knn_prob)
auc_knn = auc(fp_knn, tp_knn)

fp_log, tp_log, _ = roc_curve(y_test, y_log_prob)
auc_log = auc(fp_log, tp_log)

fp_nbc, tp_nbc, _ = roc_curve(y_test, y_nbc_prob)
auc_nbc = auc(fp_nbc, tp_nbc)

plt.figure(figsize=(10, 7))
plt.plot(fp_knn, tp_knn, label=f'KNN (AUC = {auc_knn:.4f})')
plt.plot(fp_log, tp_log, label=f'Log Regression (AUC = {auc_log:.4f})')
plt.plot(fp_nbc, tp_nbc, label=f'NBC (AUC = {auc_nbc:.4f})')

plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC-кривые для различных классификаторов')
plt.legend(loc='lower right')
plt.grid(True)
plt.show()
```

```python
from sklearn.metrics import precision_recall_curve

precision_log, recall_log, _ = precision_recall_curve(y_test, y_log_prob)
precision_knn, recall_knn, _ = precision_recall_curve(y_test, y_knn_prob)
precision_nbc, recall_nbc, _ = precision_recall_curve(y_test, y_nbc_prob)

plt.figure(figsize=(10, 7))
plt.plot(recall_log, precision_log, label=f'PRC for log regression')
plt.plot(recall_knn, precision_knn, label=f'PRC for KNN')
plt.plot(recall_nbc, precision_nbc, label=f'PRC for NBC')
plt.xlabel('Recall')
plt.ylabel('Precision')
plt.title('Precision-Recall кривая для различных классификаторов')
plt.legend(loc='lower left')
plt.grid(True)
plt.show()
```
