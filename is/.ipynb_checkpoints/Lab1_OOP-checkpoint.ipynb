"Построить три класса (базовый и 3 потомка), описывающих некоторых хищных животных (один из потомков), всеядных(второй потомок) и травоядных (третий потомок). Описать в базовом классе абстрактный метод для расчета количества и типа пищи, необходимого для пропитания животного в зоопарке.

a. Упорядочить всю последовательность животных по убыванию количества пищи. При совпадении значений – упорядочивать данные по алфавиту по имени. Вывести идентификатор животного, имя, тип и количество потребляемой пищи для всех элементов списка.

b. Вывести первые 5 имен животных из полученного в пункте а) списка.

c. Вывести последние 3 идентификатора животных из полученного в пункте а) списка.

d. Организовать запись и чтение коллекции в/из файл.

e. Организовать обработку некорректного формата входного файла."


```python
from abc import ABC, abstractmethod

class Animal(ABC):

    def __init__(self, id, name, weight, activ_level):
        self._id = id
        self._name = name
        self._weight = weight
        self._activ_level = activ_level
        self._type = ''
        self._type_food = ''
        self._count_food = 0
        self._food_coef = 0
        
    @abstractmethod
    def cuclculate_eat(self):
        pass

    def get_info(self):
        print(f'ID: {self._id}, Name: {self._name}, Type animal: {self._type}, Food count: {self._count_food} kg')
```

```python
class Predators(Animal):
    def __init__(self, id, name, weight, activ_level):
        super().__init__(id, name, weight, activ_level)
        self._type = 'Predators'
        self._type_food = 'Meet and fish'
        self._food_coef = 0.9
    
    def cuclculate_eat(self):
        if self._weight < 10:
            self._count_food = ((self._weight * 0.15 + self._activ_level * 1.5) * self._food_coef)/10
        elif self._weight < 250:
            self._count_food = ((self._weight * 0.15 + self._activ_level * 1.5) * self._food_coef)/4
        else:
            self._count_food = ((self._weight * 0.15 + self._activ_level * 1.5) * self._food_coef)/10
        self._count_food = round(self._count_food, 2)
        return self._count_food

class Herbivores(Animal):
    def __init__(self, id, name, weight, activ_level):
        super().__init__(id, name, weight, activ_level)
        self._type = 'Herbivores'
        self._type_food = 'Grass and vegetables'
        self._food_coef = 1.2
        
    def cuclculate_eat(self):
        if self._weight < 10:
            self._count_food = ((self._weight * 0.1 + self._activ_level * 1.2) * self._food_coef)/20
        elif self._weight < 350:
            self._count_food = ((self._weight * 0.1 + self._activ_level * 1.2) * self._food_coef)/5
        else:
            self._count_food = ((self._weight * 0.1 + self._activ_level * 1.2) * self._food_coef)/2
        self._count_food = round(self._count_food, 2)
        return self._count_food


class Omnivores(Animal):
    def __init__(self, id, name, weight, activ_level):
        super().__init__(id, name, weight, activ_level)
        self._type = 'Omnivores'
        self._type_food = 'Any food'
        self._food_coef = 1.0

    def cuclculate_eat(self):
        if self._weight < 10:
            self._count_food = ((self._weight * 0.12 + self._activ_level * 1.3) * self._food_coef)/10
        elif self._weight < 250:
            self._count_food = ((self._weight * 0.12 + self._activ_level * 1.3) * self._food_coef)/5
        else:
            self._count_food = ((self._weight * 0.12 + self._activ_level * 1.3) * self._food_coef)/10
        self._count_food = round(self._count_food, 2)
        return self._count_food
```

```python
lion = Predators(1, 'Lion', 235, 8) 
polar_bear = Predators(2, 'Polar bear', 600, 7)
brown_bear = Omnivores(3, 'Brown bear', 400, 6)
tiger = Predators(4, 'Tiger', 185, 8)
raccoon = Omnivores(5, 'Raccoon', 7, 6)
giraffe = Herbivores(6, 'Giraffe', 1500, 5)
elephant = Herbivores(7, 'Elephant', 4000, 3)
hare = Herbivores(8, 'Hare', 5, 9)
white_hare = Herbivores(11, 'Wight Hare', 5, 9)
zebra = Herbivores(9, 'Zebra', 320, 4)
wolf = Predators(10, 'Wolf', 45, 8)

animals = [lion, polar_bear, brown_bear, tiger, raccoon, giraffe, elephant, hare, white_hare, zebra, wolf]
```

```python
for animal in animals:
    animal.cuclculate_eat()

animals.sort(key = lambda x: (-x._count_food, x._name))

print('Животные в зоопарке: ')
for animal in animals:
    animal.get_info()
```

```python
print('The names of the first five animals: ')
for animal in animals[:5]:
    print(animal._name)
```

```python
print('The ID of the last three animals: ')
for animal in animals[8:]:
    print(animal._id)
```

```python
def write_to_txt(filename, dataset):
    with open(filename, 'w') as f:
        for data in dataset:
            f.write(f"{data._id}, {data._name}, {data._type}, {data._weight}, {data._activ_level}, {data._type_food}, {data._count_food}\n")
```

```python
def read_from_txt(filename):
    animals_zoo = []
    try:
        if not '.txt' in filename:
            raise Exception('Incorrect file format')
        with open(filename, 'r', encoding='utf-8') as f:
            for line in f:
                try:
                    animal = None
                    split = [item.strip() for item in line.split(",")]
                    if len(split) != 7:
                        raise ValueError("Incorrect string format")
                    id = int(split[0])
                    name = split[1]
                    type = split[2]
                    weight = int(split[3])
                    activ_level = int(split[4])
                    type_food = split[5]
                    count_food = float(split[6])
                    if type == 'Predators':
                        animal = Predators(id, name, weight, activ_level)
                    elif type == 'Herbivores':
                        animal = Herbivores(id, name, weight, activ_level)
                    elif type == 'Omnivores':
                        animal = Omnivores(id, name, weight, activ_level)
                    else:
                        print(f"Unknown animal type '{type}' for line")
            
                    animal._count_food = count_food
                    animals_zoo.append(animal)
                except (ValueError, IndexError) as e:
                    print(f"Error: the string '{line.strip()}' is incorrect. Reason: {e}")
        return animals_zoo
    except FileNotFoundError:
        print(f"File {filename} not found.")
    except IOError:
        print(f"File reading error {filename}.")
```

```python
write_to_txt('animals.txt', animals)
animals_zoo = read_from_txt('animals.txt')

for animal in animals_zoo:
    animal.get_info()
```
