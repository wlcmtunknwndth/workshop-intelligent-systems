## Выполнение: 

### 1. Построить онтологию для предметной области, которая указана под Вашими ФИО.

```python
class Entity:
    def __init__(self, entity_id, name):
        self.id = entity_id
        self.name = name
        self.properties = {}
        self.relationships = {}
    
    def add_property(self, key, value):
        self.properties[key] = value
    
    def add_relationship(self, relation, target):
        if relation not in self.relationships:
            self.relationships[relation] = []
        self.relationships[relation].append(target)
    
    def __str__(self):
        return f"{self.__class__.__name__}: {self.name} (ID: {self.id})"

class Researcher(Entity):
    def __init__(self, researcher_id, name, position, specialization, email):
        super().__init__(researcher_id, name)
        self.position = position
        self.specialization = specialization
        self.email = email
        self.projects = []
        self.publications = []

class Experiment(Entity):
    def __init__(self, experiment_id, title, description, start_date, status):
        super().__init__(experiment_id, title)
        self.description = description
        self.start_date = start_date
        self.status = status
        self.researchers = []
        self.equipment = []
        self.results = []

class Equipment(Entity):
    def __init__(self, equipment_id, name, equipment_type, status):
        super().__init__(equipment_id, name)
        self.type = equipment_type
        self.status = status
        self.calibration_date = None

class Publication(Entity):
    def __init__(self, pub_id, title, journal, year):
        super().__init__(pub_id, title)
        self.journal = journal
        self.year = year
        self.authors = []
        self.doi = None

class ResearchData(Entity):
    def __init__(self, data_id, name, data_type, format, size):
        super().__init__(data_id, name)
        self.data_type = data_type  # raw, processed, result
        self.format = format
        self.size = size
        self.metadata = {}

class Project(Entity):
    def __init__(self, project_id, title, funding_source, start_date):
        super().__init__(project_id, title)
        self.funding_source = funding_source
        self.start_date = start_date
        self.end_date = None
        self.researchers = []
        self.experiments = []
```

### 2. Построить базу знаний на основе этой онтологии.

```python
class VirtualResearchLab:
    """Виртуальная исследовательская лаборатория - основная база знаний"""
    
    def __init__(self):
        self.researchers = {}
        self.experiments = {}
        self.equipment = {}
        self.publications = {}
        self.datasets = {}
        self.projects = {}
        self.relationships = []
    
    def add_researcher(self, researcher):
        self.researchers[researcher.id] = researcher
    
    def add_experiment(self, experiment):
        self.experiments[experiment.id] = experiment
    
    def add_equipment(self, equipment):
        self.equipment[equipment.id] = equipment
    
    def add_publication(self, publication):
        self.publications[publication.id] = publication
    
    def add_data(self, data):
        self.datasets[data.id] = data
    
    def add_project(self, project):
        self.projects[project.id] = project
    
    def link_researcher_to_experiment(self, researcher_id, experiment_id, role="investigator"):
        """Связывает исследователя с экспериментом"""
        if researcher_id in self.researchers and experiment_id in self.experiments:
            researcher = self.researchers[researcher_id]
            experiment = self.experiments[experiment_id]
            
            researcher.add_relationship(f"works_on_{role}", experiment)
            experiment.add_relationship(f"has_{role}", researcher)
            
            researcher.projects.append(experiment)
            experiment.researchers.append(researcher)
            
            self.relationships.append({
                'type': f'researcher_{role}_experiment',
                'from': researcher_id,
                'to': experiment_id
            })
    
    def link_experiment_to_equipment(self, experiment_id, equipment_id):
        """Связывает эксперимент с оборудованием"""
        if experiment_id in self.experiments and equipment_id in self.equipment:
            experiment = self.experiments[experiment_id]
            equipment = self.equipment[equipment_id]
            
            experiment.add_relationship("uses_equipment", equipment)
            equipment.add_relationship("used_in", experiment)
            
            experiment.equipment.append(equipment)
            
            self.relationships.append({
                'type': 'experiment_uses_equipment',
                'from': experiment_id,
                'to': equipment_id
            })
    
    def link_publication_to_authors(self, publication_id, researcher_ids):
        """Связывает публикацию с авторами"""
        if publication_id in self.publications:
            publication = self.publications[publication_id]
            
            for researcher_id in researcher_ids:
                if researcher_id in self.researchers:
                    researcher = self.researchers[researcher_id]
                    publication.authors.append(researcher)
                    researcher.publications.append(publication)
                    
                    publication.add_relationship("has_author", researcher)
                    researcher.add_relationship("author_of", publication)
                    
                    self.relationships.append({
                        'type': 'author_publication',
                        'from': researcher_id,
                        'to': publication_id
                    })
    
    def link_experiment_to_data(self, experiment_id, data_id, data_type="result"):
        """Связывает эксперимент с данными"""
        if experiment_id in self.experiments and data_id in self.datasets:
            experiment = self.experiments[experiment_id]
            data = self.datasets[data_id]
            
            experiment.add_relationship(f"produces_{data_type}_data", data)
            data.add_relationship(f"produced_by_{data_type}", experiment)
            
            experiment.results.append(data)
            
            self.relationships.append({
                'type': f'experiment_produces_{data_type}',
                'from': experiment_id,
                'to': data_id
            })
    
    def find_researchers_by_specialization(self, specialization):
        """Найти исследователей по специализации"""
        return [r for r in self.researchers.values() 
                if r.specialization == specialization]
    
    def find_experiments_by_status(self, status):
        """Найти эксперименты по статусу"""
        return [e for e in self.experiments.values() 
                if e.status == status]
    
    def find_equipment_by_type(self, equipment_type):
        """Найти оборудование по типу"""
        return [eq for eq in self.equipment.values() 
                if eq.type == equipment_type]
    
    def get_researcher_publications(self, researcher_id):
        """Получить публикации исследователя"""
        if researcher_id in self.researchers:
            return self.researchers[researcher_id].publications
        return []
    
    def get_experiment_participants(self, experiment_id):
        """Получить участников эксперимента"""
        if experiment_id in self.experiments:
            return self.experiments[experiment_id].researchers
        return []
    
    def get_equipment_usage(self, equipment_id):
        """Получить эксперименты, где использовалось оборудование"""
        if equipment_id in self.equipment:
            equipment = self.equipment[equipment_id]
            experiments = []
            for exp in self.experiments.values():
                if equipment in exp.equipment:
                    experiments.append(exp)
            return experiments
        return []
    
    def get_data_by_experiment(self, experiment_id):
        """Получить данные, произведенные экспериментом"""
        if experiment_id in self.experiments:
            return self.experiments[experiment_id].results
        return []
    
    def visualize_relationships(self):
        """Визуализация связей в лаборатории (текстовый вывод)"""
        print("\n" + "="*60)
        print("СТРУКТУРА ВИРТУАЛЬНОЙ ИССЛЕДОВАТЕЛЬСКОЙ ЛАБОРАТОРИИ")
        print("="*60)
        
        print("\nИССЛЕДОВАТЕЛИ:")
        for researcher in self.researchers.values():
            print(f"  - {researcher.name} ({researcher.position})")
            if researcher.relationships:
                print("    Связи:", ", ".join(researcher.relationships.keys()))
        
        print("\nЭКСПЕРИМЕНТЫ:")
        for experiment in self.experiments.values():
            print(f"  - {experiment.name} ({experiment.status})")
            print(f"    Участники: {[r.name for r in experiment.researchers]}")
        
        print("\nОБОРУДОВАНИЕ:")
        for equip in self.equipment.values():
            print(f"  - {equip.name} ({equip.type}, статус: {equip.status})")
        
        print("\nПУБЛИКАЦИИ:")
        for pub in self.publications.values():
            print(f"  - {pub.name} ({pub.year}, {pub.journal})")
            print(f"    Авторы: {[a.name for a in pub.authors]}")
        
        print("\nОБЩИЕ СВЯЗИ:")
        for rel in self.relationships:
            print(f"  {rel['type']}: {rel['from']} -> {rel['to']}")
```

### 3. Реализуйте приложение, которое отвечает на вопросы при создании онтологии используя построенную базу знаний.

```python
class ResearchLabAssistant:
    """Ассистент для вопросно-ответной системы"""
    
    def __init__(self, lab):
        self.lab = lab
        self.commands = {
            'поиск исследователей': self.search_researchers,
            'эксперименты по статусу': self.search_experiments_by_status,
            'оборудование по типу': self.search_equipment,
            'публикации автора': self.get_author_publications,
            'участники эксперимента': self.get_experiment_participants,
            'использование оборудования': self.get_equipment_usage,
            'данные эксперимента': self.get_experiment_data,
            'показать структуру': self.show_structure,
            'помощь': self.show_help,
            'выход': exit
        }
    
    def search_researchers(self):
        specialization = input("Введите специализацию для поиска: ")
        researchers = self.lab.find_researchers_by_specialization(specialization)
        if researchers:
            print(f"\nНайдено {len(researchers)} исследователей:")
            for r in researchers:
                print(f"  - {r.name}, {r.position}, email: {r.email}")
        else:
            print("Исследователи не найдены.")
    
    def search_experiments_by_status(self):
        status = input("Введите статус эксперимента (active/completed/planned): ")
        experiments = self.lab.find_experiments_by_status(status)
        if experiments:
            print(f"\nНайдено {len(experiments)} экспериментов:")
            for e in experiments:
                print(f"  - {e.name}: {e.description}")
                print(f"    Начало: {e.start_date}, Статус: {e.status}")
        else:
            print("Эксперименты не найдены.")
    
    def search_equipment(self):
        eq_type = input("Введите тип оборудования: ")
        equipment = self.lab.find_equipment_by_type(eq_type)
        if equipment:
            print(f"\nНайдено {len(equipment)} единиц оборудования:")
            for eq in equipment:
                print(f"  - {eq.name}, статус: {eq.status}")
        else:
            print("Оборудование не найдено.")
    
    def get_author_publications(self):
        researcher_id = input("Введите ID исследователя: ")
        publications = self.lab.get_researcher_publications(researcher_id)
        if publications:
            print(f"\nПубликации исследователя:")
            for pub in publications:
                print(f"  - {pub.name} ({pub.year}, {pub.journal})")
        else:
            print("Публикации не найдены.")
    
    def get_experiment_participants(self):
        experiment_id = input("Введите ID эксперимента: ")
        participants = self.lab.get_experiment_participants(experiment_id)
        if participants:
            print(f"\nУчастники эксперимента:")
            for p in participants:
                print(f"  - {p.name} ({p.position}, {p.specialization})")
        else:
            print("Участники не найдены.")
    
    def get_equipment_usage(self):
        equipment_id = input("Введите ID оборудования: ")
        experiments = self.lab.get_equipment_usage(equipment_id)
        if experiments:
            print(f"\nОборудование используется в экспериментах:")
            for exp in experiments:
                print(f"  - {exp.name} ({exp.status})")
        else:
            print("Использование оборудования не найдено.")
    
    def get_experiment_data(self):
        experiment_id = input("Введите ID эксперимента: ")
        datasets = self.lab.get_data_by_experiment(experiment_id)
        if datasets:
            print(f"\nДанные эксперимента:")
            for data in datasets:
                print(f"  - {data.name} ({data.data_type}, {data.format}, {data.size} MB)")
        else:
            print("Данные не найдены.")
    
    def show_structure(self):
        self.lab.visualize_relationships()
    
    def show_help(self):
        print("\nДоступные команды:")
        for cmd in self.commands.keys():
            print(f"  - {cmd}")
        print()
    
    def run(self):
        print("\n" + "="*60)
        print("АССИСТЕНТ ВИРТУАЛЬНОЙ ИССЛЕДОВАТЕЛЬСКОЙ ЛАБОРАТОРИИ")
        print("="*60)
        
        self.show_help()
        
        while True:
            try:
                command = input("\nВведите команду: ").lower().strip()
                
                if command in self.commands:
                    self.commands[command]()
                else:
                    print("Неизвестная команда. Введите 'помощь' для списка команд.")
            except KeyboardInterrupt:
                print("\n\nРабота завершена.")
                break
            except Exception as e:
                print(f"Ошибка: {e}")
```

### 4. Тестирование

```python
def create_sample_lab():
    lab = VirtualResearchLab()
    
    # Создаем исследователей
    researcher1 = Researcher("RES001", "Петров Артем Евгеньевич", 
                            "Старший научный сотрудник", 
                            "Наноматериалы", "artem@vrlab.ru")
    researcher2 = Researcher("RES002", "Иванова Мария Сергеевна", 
                            "Младший научный сотрудник", 
                            "Биоинформатика", "maria@vrlab.ru")
    researcher3 = Researcher("RES003", "Сидоров Павел Игоревич", 
                            "Инженер-исследователь", 
                            "Анализ данных", "pavel@vrlab.ru")
    
    lab.add_researcher(researcher1)
    lab.add_researcher(researcher2)
    lab.add_researcher(researcher3)
    
    # Создаем оборудование
    microscope = Equipment("EQ001", "Электронный микроскоп JEOL", 
                          "Микроскопия", "operational")
    spectrometer = Equipment("EQ002", "Масс-спектрометр Thermo", 
                           "Спектрометрия", "calibration_required")
    sequencer = Equipment("EQ003", "Секвенатор Illumina", 
                         "Геномика", "operational")
    
    lab.add_equipment(microscope)
    lab.add_equipment(spectrometer)
    lab.add_equipment(sequencer)
    
    # Создаем эксперименты
    exp1 = Experiment("EXP001", "Синтез наночастиц золота", 
                     "Исследование методов синтеза наночастиц", 
                     "2025-01-15", "completed")
    exp2 = Experiment("EXP002", "Анализ генома бактерии", 
                     "Секвенирование и анализ генома E. coli", 
                     "2025-02-01", "active")
    exp3 = Experiment("EXP003", "Исследование каталитических свойств", 
                     "Тестирование катализаторов для реакций", 
                     "2025-03-10", "planned")
    
    lab.add_experiment(exp1)
    lab.add_experiment(exp2)
    lab.add_experiment(exp3)
    
    # Создаем публикации
    pub1 = Publication("PUB001", "Новый метод синтеза наноматериалов", 
                      "Journal of Nanoscience", 2024)
    pub2 = Publication("PUB002", "Анализ геномных данных бактерий", 
                      "Bioinformatics Journal", 2024)
    
    lab.add_publication(pub1)
    lab.add_publication(pub2)
    
    # Создаем данные
    data1 = ResearchData("DATA001", "Спектры наночастиц", 
                        "processed", "CSV", 150)
    data2 = ResearchData("DATA002", "Геномные последовательности", 
                        "raw", "FASTQ", 2500)
    
    lab.add_data(data1)
    lab.add_data(data2)
    
    # Создаем связи
    lab.link_researcher_to_experiment("RES001", "EXP001", "lead")
    lab.link_researcher_to_experiment("RES002", "EXP001", "assistant")
    lab.link_researcher_to_experiment("RES002", "EXP002", "lead")
    lab.link_researcher_to_experiment("RES003", "EXP002", "analyst")
    
    lab.link_experiment_to_equipment("EXP001", "EQ001")
    lab.link_experiment_to_equipment("EXP001", "EQ002")
    lab.link_experiment_to_equipment("EXP002", "EQ003")
    
    lab.link_publication_to_authors("PUB001", ["RES001", "RES002"])
    lab.link_publication_to_authors("PUB002", ["RES002", "RES003"])
    
    lab.link_experiment_to_data("EXP001", "DATA001", "processed")
    lab.link_experiment_to_data("EXP002", "DATA002", "raw")
    
    return lab
```

```python
lab = create_sample_lab()

assistant = ResearchLabAssistant(lab)
assistant.run()
```
